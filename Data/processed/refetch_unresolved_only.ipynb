# Refetch Unresolved Sequences Only

This notebook reruns sequence fetch only for the current human unresolved IDs in `Data/output_data/unresolved_ids.csv`.
It does not rebuild the full website pipeline.

```python
from pathlib import Path
import sys

def find_project_root(marker_rel: Path = Path("Data/raw/website_alldata.csv")) -> Path:
    cwd = Path.cwd().resolve()
    for base in (cwd, *cwd.parents):
        if (base / marker_rel).exists():
            return base
    raise FileNotFoundError(f"Could not locate project root containing {marker_rel}")

PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from scripts.refetch_unresolved_sequences import main

PROJECT_ROOT
```

```python
# Set max_ids to a small number for a quick test, or None for all current human unresolved IDs.
MAX_IDS = None
TIMEOUT = 10
SLEEP_SEC = 0.05
PROGRESS_EVERY = 25

argv = [
    "--timeout", str(TIMEOUT),
    "--sleep_sec", str(SLEEP_SEC),
    "--progress_every", str(PROGRESS_EVERY),
]
if MAX_IDS is not None:
    argv.extend(["--max_ids", str(MAX_IDS)])

main(argv)
```
